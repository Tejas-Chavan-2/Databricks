#Manged table vs External Delta Table


## Create Database

```python
%sql
create database ProductDB
```

###Create Managed Table

```python
%sql
use ProductDB;
create table ProductTable(
  id int,
  name string,
  price int
)
using delta 
```

```python
%sql
insert into ProductDB.ProductTable(id,name,price) values(1,'Cap',100),(2,'Shirt',600)
```

```python
%sql
select * from ProductDB.ProductTable;
```

```python
%sql
insert into ProductDB.ProductTable(id,name,price) values(6,'Shoes',1500),(4,'Phone',20060),(12,'Ball',20),(15,'Kit',2700),(20,'Lock',350)
```

```python
%sql
select * from ProductDB.ProductTable order by id;
```

```python
%sql
update ProductDB.ProductTable set price=400 where id=12;
select * from ProductDB.ProductTable order by id;
```

```python
%sql
delete from ProductDB.ProductTable where id=12;
select * from ProductDB.ProductTable order by id;
```
